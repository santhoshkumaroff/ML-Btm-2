```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
import pandas as pd
```

```python
df = pd.DataFrame({'message':['ML is Subset of AI','AI is Superset of Both ML and DL ML and DL','GenAI is used to create or generate any type of content']})
```

```python
df['data'] = df['message']
df['new'] =df['data']+' '+df['message']

df
```

```python
vector = CountVectorizer(min_df=2)
val = vector.fit_transform(df['message'])
check = pd.DataFrame(val.toarray(),columns=vector.get_feature_names_out())
check
```

```python
vector = CountVectorizer()
val = vector.fit_transform(df['message'])
check = pd.DataFrame(val.toarray(),columns=vector.get_feature_names_out())
check
```

```python
vector = CountVectorizer(min_df=2,max_df=3)
val = vector.fit_transform(df['message'])
check = pd.DataFrame(val.toarray(),columns=vector.get_feature_names_out())
check
```

```python
vector = CountVectorizer()
val = vector.fit_transform(df['message'])
check = pd.DataFrame(val.toarray(),columns=vector.get_feature_names_out())
check
```

```python
vector = TfidfVectorizer(stop_words='english',smooth_idf=True)
val = vector.fit_transform(df['message'])
check = pd.DataFrame(val.toarray(),columns=vector.get_feature_names_out())
check
```
